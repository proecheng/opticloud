# 线性规划教学模式

这个 Notebook 配合 OptiCloud `mode=teaching` 使用，展示一个最小 LP：定义决策变量、线性目标函数和线性约束，然后提交到 `/v1/optimizations?mode=teaching`。

教学模式会返回原理讲解、50% Credits 折扣资格和本 Notebook 的 Colab 链接。

## 模型

示例问题：最小化 `x1 + x2`，约束为 `x1 + x2 <= 10`。课堂中可以把 `x1`、`x2` 解释为两类资源或两条路线的投入量。

```python
payload = {
    "task_type": "lp",
    "minimize": {"c": [1.0, 1.0]},
    "st": {"A": [[1.0, 1.0]], "b": [10.0]},
}

{
    "endpoint": "/v1/optimizations?mode=teaching",
    "payload": payload,
}
```

## 解释结果

返回中的 `teaching.principle_explanation` 给出 LP 建模步骤和局限；`teaching.credits_discount.discount_multiplier = 0.5` 表示教学模式的 Credits 折扣资格。真实扣费仍以后端 billing finalize 记录为准。
